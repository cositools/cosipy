```python
from pathlib import Path
from cosipy.util import fetch_wasabi_file

# Define the local data directory (created if it doesn't exist)
data_path = Path("./dc2")
data_path.mkdir(exist_ok=True)

# Define both the Wasabi key (remote path) and local file path
# We'll work directly with the gzipped FITS file
local_gz = data_path / "Crab_DC2_3months_unbinned_data.fits.gz"
wasabi_key = "COSI-SMEX/DC2/Data/Sources/Crab_DC2_3months_unbinned_data.fits.gz"

# Check if the file already exists locally — if yes, reuse it.
# Otherwise, download it from the Wasabi cloud storage.
if local_gz.exists():
    print(f"File already exists, using local copy: {local_gz}")
else:
    fetch_wasabi_file(wasabi_key, output=str(local_gz))
    print(f"Fetched from Wasabi: {local_gz}")
```

```python
from cosipy_pulsar import PulsarAnalyzer

local_fits = local_gz
cfg_path = "config.yaml"
an = PulsarAnalyzer.from_yaml(cfg_path)
an.cfg.local_fits = str(local_fits)  # point to the fetched file

fig = an.run(show=True)
```
